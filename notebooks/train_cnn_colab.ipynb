# Train the SoundView CNN in Colab
Select a GPU runtime, then run the cells in order. The setup cell downloads the project and ESC-50 into this temporary Colab session.

```python
import subprocess
import sys
from pathlib import Path

project_dir = Path('/content/soundview')
data_dir = Path('/content/datasets/ESC-50')

# Colab starts with a temporary disk, so download the code and audio here.
if not project_dir.exists():
    subprocess.run(['git', 'clone', 'https://github.com/AndreuLopezz/12607559_SoundView-environmental-sound-recognition.git', str(project_dir)], check=True)
if not data_dir.exists():
    data_dir.parent.mkdir(parents=True, exist_ok=True)
    subprocess.run(['git', 'clone', '--depth', '1', 'https://github.com/karolpiczak/ESC-50.git', str(data_dir)], check=True)

subprocess.run([sys.executable, '-m', 'pip', 'install', '-r', str(project_dir / 'requirements.txt')], check=True)
sys.path.insert(0, str(project_dir))
```

## Settings
Change the values below before training. The output folder is temporary: copy the results out of the Colab runtime before it disconnects. Use a new folder for each experiment.

```python
TEST_FOLD = 5
EPOCHS = 40
BATCH_SIZE = 32
LEARNING_RATE = 0.001
PATIENCE = 5
SEED = 42
OUTPUT_DIR = Path('/content/runs/cnn_fold5_40epochs')
```

```python
from soundview.train_cnn import train_cnn

train_cnn(
    data_dir=data_dir,
    output_dir=OUTPUT_DIR,
    test_fold=TEST_FOLD,
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    learning_rate=LEARNING_RATE,
    patience=PATIENCE,
    seed=SEED,
)
print(f'Retrieve {OUTPUT_DIR / "best_cnn.pt"} and {OUTPUT_DIR / "training_summary.json"} before disconnecting.')
```

```python
from pathlib import Path
from shutil import copy2
from google.colab import drive

drive.mount("/content/drive")

origen = OUTPUT_DIR
destino = Path("/content/drive/MyDrive/SoundView") / OUTPUT_DIR.name
destino.mkdir(parents=True, exist_ok=True)

for nombre in ("best_cnn.pt", "training_summary.json"):
    archivo = origen / nombre
    copia = copy2(archivo, destino / nombre)
    print(f"Guardado: {copia}")
```
