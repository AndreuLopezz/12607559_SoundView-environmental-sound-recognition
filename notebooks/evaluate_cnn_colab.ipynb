# Evaluate the saved CNN on fold 5
Run this after choosing the checkpoint using validation results. The test fold is measured once; do not use its score to change the model settings. Select a Colab GPU runtime and run the cells in order.

```python
import subprocess
import sys
from pathlib import Path

from google.colab import drive

project_dir = Path('/content/soundview')
data_dir = Path('/content/datasets/ESC-50')

# Refresh the code if this Colab session was used for training earlier.
if project_dir.exists():
    subprocess.run(['git', '-C', str(project_dir), 'pull', '--ff-only'], check=True)
else:
    subprocess.run(['git', 'clone', 'https://github.com/AndreuLopezz/12607559_SoundView-environmental-sound-recognition.git', str(project_dir)], check=True)
if not data_dir.exists():
    data_dir.parent.mkdir(parents=True, exist_ok=True)
    subprocess.run(['git', 'clone', '--depth', '1', 'https://github.com/karolpiczak/ESC-50.git', str(data_dir)], check=True)

subprocess.run([sys.executable, '-m', 'pip', 'install', '-r', str(project_dir / 'requirements.txt')], check=True)
sys.path.insert(0, str(project_dir))
drive.mount('/content/drive')
```

```python
from soundview.evaluate_cnn import evaluate_cnn

TEST_FOLD = 5
BATCH_SIZE = 32
RUN_DIR = Path('/content/drive/MyDrive/SoundView/cnn_fold5_40epochs')
CHECKPOINT = RUN_DIR / 'best_cnn.pt'
REPORT = RUN_DIR / 'test_evaluation.json'

result = evaluate_cnn(
    data_dir=data_dir,
    checkpoint_path=CHECKPOINT,
    output_path=REPORT,
    test_fold=TEST_FOLD,
    batch_size=BATCH_SIZE,
)
print(f"Test accuracy: {result['accuracy']:.4f}")
print(f"Test macro-F1: {result['macro_f1']:.4f}")
print(f"Mean model inference: {result['inference_ms_per_clip']:.2f} ms per clip")
print(f'Report saved to {REPORT}')
```
